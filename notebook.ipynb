```python
import random
import matplotlib.pyplot as plt

random.seed(10)

horas = [1, 2, 3, 4, 5, 6, 7, 8, 9]
notas = [min(10, h + random.uniform(0, 2)) for h in horas]

print("Horas de estudio:", horas)
print("Notas:", [round(n, 2) for n in notas])
```

```python
# La IA aprende una recta: nota = pendiente * horas + inicio

media_horas = sum(horas) / len(horas)
media_notas = sum(notas) / len(notas)

pendiente = sum(
    (x - media_horas) * (y - media_notas)
    for x, y in zip(horas, notas)
) / sum((x - media_horas) ** 2 for x in horas)

inicio = media_notas - pendiente * media_horas

print(f"Pendiente aprendida: {pendiente:.2f}")
print(f"Valor inicial: {inicio:.2f}")
```

```python
def predecir_nota(horas_estudio):
    return min(10, max(0, pendiente * horas_estudio + inicio))

horas_nuevas = 7.5
prediccion = predecir_nota(horas_nuevas)

print(f"Si estudias {horas_nuevas} horas, la nota estimada es: {prediccion:.2f}")
```

```python
plt.scatter(horas, notas, color="royalblue", label="Datos reales")

linea_x = list(range(1, 11))
linea_y = [predecir_nota(x) for x in linea_x]

plt.plot(linea_x, linea_y, color="crimson", label="Predicción de la IA")
plt.scatter(horas_nuevas, prediccion, color="green", s=120, label="Nueva predicción")

plt.title("Mini IA: horas de estudio y notas")
plt.xlabel("Horas de estudio")
plt.ylabel("Nota")
plt.ylim(0, 11)
plt.grid(True)
plt.legend()
plt.show()
```
